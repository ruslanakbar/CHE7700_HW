# Prime number logic

```python
# (a) Define prime check function
def is_prime(num):
    if num <= 1:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

# (b) & (c) General function for primes below n
def find_primes_below(n):
    return [i for i in range(2, n) if is_prime(i)]

# (d) Call for n = 1000
primes_100 = find_primes_below(100)
primes_1000 = find_primes_below(1000)
print(f"Primes below 100: {primes_100}")
```

# NumPy Array Generation

```python
import numpy as np

# (a) Function for 3D array
def create_3d_array(n, data_type='zeros'):
    if data_type == 'zeros':
        return np.zeros((n, 3, 3))
    elif data_type == 'random':
        return np.random.randint(1, 101, size=(n, 3, 3))

# (b) n=10, zeros
zeros_array = create_3d_array(10, 'zeros')

# (c) n=5, random 1-100
random_array = create_3d_array(5, 'random')
```

```python
A = np.array([[3, 4, -5], 
              [0, 5, 1], 
              [1, 3, 6]])
B = np.array([10, 8, -5])

solution = np.linalg.solve(A, B)
print(f"a: {solution[0]}, b: {solution[1]}, c: {solution[2]}")
```

```python
import matplotlib.pyplot as plt

# (a) Define and plot
x = np.linspace(-5, 5, 400)
y = 0.1 * x**3 - x**2 + 5

plt.plot(x, y, label='$y = 0.1x^3 - x^2 + 5$')

# (b) Find Max/Min in this range
y_max, y_min = np.max(y), np.min(y)
x_max, x_min = x[np.argmax(y)], x[np.argmin(y)]

plt.scatter([x_max, x_min], [y_max, y_min], color='red')
plt.text(x_max, y_max, f' Max ({x_max:.2f}, {y_max:.2f})')
plt.text(x_min, y_min, f' Min ({x_min:.2f}, {y_min:.2f})')

plt.grid(True)
plt.legend()
plt.show()
```
